Challenge 1

```python
total = 0
count = 0
average = 0

while True:
    line = input("Enter a number: ")
    if line == 'done':
        break
    try:
        number = float(line)
    except ValueError:
        print("Invalid input")
        continue

    total = total + number
    count = count + 1

if count > 0:
    average = total / count

print(total, count, average)
```

Challenge 2

```python
max_num = None
min_num = None

while True:
    line = input("Enter a number: ")
    if line == 'done':
        break
    try:
        number = float(line)
    except ValueError:
        print("Invalid input")
        continue

    if max_num is None or number > max_num:
        max_num = number
    if min_num is None or number < min_num:
        min_num = number

print("Maximum:", max_num)
print("Minimum:", min_num)
```
